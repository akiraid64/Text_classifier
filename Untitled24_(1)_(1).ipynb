```python
import pandas as pd
from sklearn.datasets import fetch_20newsgroups

# Define categories
categories = ['rec.sport.baseball', 'sci.med']

# Fetch train & test splits
train_data = fetch_20newsgroups(subset='train',
                                categories=categories,
                                shuffle=True,
                                random_state=42)
test_data  = fetch_20newsgroups(subset='test',
                                categories=categories,
                                shuffle=True,
                                random_state=42)

# Create DataFrames
df_train = pd.DataFrame({
    'text':  train_data.data,
    'label': train_data.target
})

df_test = pd.DataFrame({
    'text':  test_data.data,
    'label': test_data.target
})

print("Train shape:", df_train.shape)
print("Test shape: ", df_test.shape)
```

```python
from datasets import Dataset

# Check if '__index_level_0__' column exists before removing it
if '__index_level_0__' in df_train.columns:
    train_ds = Dataset.from_pandas(df_train).remove_columns("__index_level_0__")
else:
    train_ds = Dataset.from_pandas(df_train)

if '__index_level_0__' in df_test.columns:
    test_ds = Dataset.from_pandas(df_test).remove_columns("__index_level_0__")
else:
    test_ds = Dataset.from_pandas(df_test)

print(train_ds)
print(test_ds)
```

```python
from transformers import DistilBertTokenizerFast

tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")

def tokenize_batch(batch):
    return tokenizer(batch["text"],
                     padding="max_length",
                     truncation=True,
                     max_length=256)

# apply
train_ds = train_ds.map(tokenize_batch, batched=True)
test_ds  = test_ds.map(tokenize_batch,  batched=True)

# keep only the inputs + label
train_ds.set_format(type="torch",
                    columns=["input_ids", "attention_mask", "label"])
test_ds.set_format( type="torch",
                    columns=["input_ids", "attention_mask", "label"])
```

```python
from transformers import DistilBertForSequenceClassification

model = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)
```

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./results-20ng",
    # Replace 'evaluation_strategy' with 'eval_strategy'
    eval_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    learning_rate=2e-5,
    weight_decay=0.01,
    logging_steps=50,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds.shuffle(seed=42),
    eval_dataset=test_ds
)
```

```python
trainer.train()
```

```python
import joblib
import torch

class NewsClassifier:
    def __init__(self, model, tokenizer):
        self.model = model
        self.tokenizer = tokenizer

    def predict(self, texts):
        self.model.eval()
        tokens = self.tokenizer(texts, padding=True, truncation=True, return_tensors="pt")
        with torch.no_grad():
            outputs = self.model(**tokens)
        predictions = torch.argmax(outputs.logits, dim=1)
        return predictions.numpy()

# Save Hugging Face format
model.save_pretrained("./distilbert-news-model")
tokenizer.save_pretrained("./distilbert-news-model")

# Wrap and save as a .pkl file
classifier = NewsClassifier(model, tokenizer)
joblib.dump(classifier, "news_classifier.pkl")
```

```python
from sklearn.metrics import classification_report, accuracy_score

# Get predictions on the test set
predictions = trainer.predict(test_ds)
pred_labels = predictions.predictions.argmax(axis=1)
true_labels = predictions.label_ids

# Accuracy
acc = accuracy_score(true_labels, pred_labels)
print(f"\nTest Accuracy: {acc:.4f}\n")

# Detailed report
print("Classification Report:\n")
print(classification_report(true_labels, pred_labels, target_names=["sports", "medical"]))
```

```python
import torch

def classify_texts_colab(texts, model, tokenizer):
    """
    Classify input texts using DistilBERT model.

    Args:
        texts (list or pd.Series): Input sentences.
        model: Trained DistilBERT model.
        tokenizer: DistilBERT tokenizer.

    Returns:
        List of predicted labels: 'medical' or 'sports'
    """
    if isinstance(texts, pd.Series):
        texts = texts.tolist()

    model.eval()
    inputs = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs)

    preds = torch.argmax(outputs.logits, dim=1).tolist()
    label_map = {0: "sports", 1: "medical"}

    return [label_map[p] for p in preds]
```

```python
import torch

def classify_texts_colab(texts, model, tokenizer):
    """
    Classify input texts using DistilBERT model.

    Args:
        texts (list or pd.Series): Input sentences.
        model: Trained DistilBERT model.
        tokenizer: DistilBERT tokenizer.

    Returns:
        List of predicted labels: 'medical' or 'sports'
    """
    if isinstance(texts, pd.Series):
        texts = texts.tolist()

    model.eval()
    inputs = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")

    # Move inputs to the same device as the model
    inputs = {k: v.to(model.device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    preds = torch.argmax(outputs.logits, dim=1).tolist()
    label_map = {0: "sports", 1: "medical"}

    return [label_map[p] for p in preds]
```

```python
# Single or multiple sentences
my_inputs = [
    "The team won their final match and secured the championship.",
    "Doctors discovered a new method to treat heart disease."
]

predictions = classify_texts_colab(my_inputs, model, tokenizer)

for text, label in zip(my_inputs, predictions):
    print(f"\"{text}\"\n→ Predicted: {label}\n")
```

```python
# ─── Setup & Cache Cleanup ────────────────────────────────────────
import os, gc
import torch
import joblib

# reduce fragmentation
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "max_split_size_mb:128"
torch.cuda.empty_cache()
gc.collect()

# ─── Load Your Pickle Classifier ─────────────────────────────────
classifier = joblib.load("news_classifier.pkl")

# ─── Device Selection + Fallback ─────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    classifier.model.to(device)
    # cut memory in half on GPU
    if device.type == "cuda":
        classifier.model.half()
except RuntimeError as e:
    print("→ GPU OOM, falling back to CPU:", e)
    torch.cuda.empty_cache()
    device = torch.device("cpu")
    classifier.model.to(device)

# ─── Inference Function ─────────────────────────────────────────
label_map = {0: "sports", 1: "medical"}  # adjust to your classes

def classify(text):
    # tokenize + send to correct device
    inputs = classifier.tokenizer(
        [text],
        padding=True,
        truncation=True,
        max_length=256,
        return_tensors="pt"
    )
    inputs = {k: v.to(device) for k, v in inputs.items()}
    classifier.model.eval()
    with torch.no_grad():
        logits = classifier.model(**inputs).logits
    pred = torch.argmax(logits, dim=-1).item()
    return label_map.get(pred, str(pred))

# ─── Gradio Interface ────────────────────────────────────────────
import gradio as gr

iface = gr.Interface(
    fn=classify,
    inputs=gr.Textbox(lines=4, placeholder="Paste your news text…"),
    outputs="text",
    title="📰 Efficient News Classifier",
    description="Loads whatever GPU memory is free, falls back to CPU on OOM, and uses fp16 on CUDA for inference."
)

iface.launch()
```
