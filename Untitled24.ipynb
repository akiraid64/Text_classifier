```python
import pandas as pd
from sklearn.datasets import fetch_20newsgroups

# Define categories
categories = ['rec.sport.baseball', 'sci.med']

# Fetch train & test splits
train_data = fetch_20newsgroups(subset='train',
                                categories=categories,
                                shuffle=True,
                                random_state=42)
test_data  = fetch_20newsgroups(subset='test',
                                categories=categories,
                                shuffle=True,
                                random_state=42)

# Create DataFrames
df_train = pd.DataFrame({
    'text':  train_data.data,
    'label': train_data.target
})

df_test = pd.DataFrame({
    'text':  test_data.data,
    'label': test_data.target
})

print("Train shape:", df_train.shape)
print("Test shape: ", df_test.shape)
```

```python
from datasets import Dataset

# Check if '__index_level_0__' column exists before removing it
if '__index_level_0__' in df_train.columns:
    train_ds = Dataset.from_pandas(df_train).remove_columns("__index_level_0__")
else:
    train_ds = Dataset.from_pandas(df_train)

if '__index_level_0__' in df_test.columns:
    test_ds = Dataset.from_pandas(df_test).remove_columns("__index_level_0__")
else:
    test_ds = Dataset.from_pandas(df_test)

print(train_ds)
print(test_ds)
```

```python
from transformers import DistilBertTokenizerFast

tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")

def tokenize_batch(batch):
    return tokenizer(batch["text"],
                     padding="max_length",
                     truncation=True,
                     max_length=256)

# apply
train_ds = train_ds.map(tokenize_batch, batched=True)
test_ds  = test_ds.map(tokenize_batch,  batched=True)

# keep only the inputs + label
train_ds.set_format(type="torch",
                    columns=["input_ids", "attention_mask", "label"])
test_ds.set_format( type="torch",
                    columns=["input_ids", "attention_mask", "label"])
```

```python
from transformers import DistilBertForSequenceClassification

model = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)
```

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./results-20ng",
    # Replace 'evaluation_strategy' with 'eval_strategy'
    eval_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    learning_rate=2e-5,
    weight_decay=0.01,
    logging_steps=50,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds.shuffle(seed=42),
    eval_dataset=test_ds
)
```

```python
trainer.train()
```
